```python
transaksi = [
    {
        "jenis": "masuk",
        "keterangan": "Modal Awal",
        "jumlah": 5000000
    },
    {
        "jenis": "masuk",
        "keterangan": "Penjualan Kopi",
        "jumlah": 150000
    },
    {
        "jenis": "keluar",
        "keterangan": "Beli Gula",
        "jumlah": 50000
    },
    {
        "jenis": "keluar",
        "keterangan": "Beli Susu",
        "jumlah": 75000
    },
    {
        "jenis": "masuk",
        "keterangan": "Penjualan Roti",
        "jumlah": 200000
    }
]

def tampil_keluar(transaksi):
    hasil = ""
    total = 0
    for i in transaksi:
        if i["jenis"] == "keluar":
           print(f"Pengeluaran : {i['keterangan']} : Rp{i['jumlah']}")
           total += i["jumlah"]
           print(f"Total Pengeluaran : {total}")
    return hasil
tampil_keluar(transaksi)
print("Program Selesai")
```

```python
transaksi = [
    {"jenis": "masuk", "jumlah": 500000},
    {"jenis": "keluar", "jumlah": 120000},
    {"jenis": "masuk", "jumlah": 300000},
    {"jenis": "keluar", "jumlah": 50000},
    {"jenis": "bonus", "jumlah": 1000}
]

def hitung_saldo(transaksi):
    Pemasukan = 0
    Pengeluaran=0
    saldo = 0
    for i in transaksi :
        if i["jenis"]=="masuk":
            Pemasukan += i["jumlah"]
            saldo += i["jumlah"]
        elif i["jenis"] == "keluar" :
            Pengeluaran += i["jumlah"]
            saldo -= i["jumlah"]
        else :
            print("transaksi tidak valid")
            print(f"{i["jenis"]} : {i["jumlah"]}")
            continue
    return saldo, Pemasukan, Pengeluaran
saldo_akhir, total_masuk, total_keluar=hitung_saldo(transaksi)

print("-"*20)
print("Pemasukan:", total_masuk)
print("Pengeluaran:", total_keluar)
print(f"Saldo Akhir : {saldo_akhir}")
print("Program Selesai")
print("-"*20)
```

```python
transaksi = [
 {"jenis":"keluar","jumlah":120000,"kategori":"belanja"},
 {"jenis":"keluar","jumlah":50000,"kategori":"transport"},
 {"jenis":"keluar","jumlah":80000,"kategori":"belanja"},
 {"jenis":"keluar","jumlah":70000,"kategori":"listrik"}
]

laporan = {}

for i in transaksi:
    kategori = i["kategori"]
    jumlah = i["jumlah"]
    if kategori not in laporan :
        laporan[kategori] =0
    laporan[kategori] += jumlah
    
print("--- Laporan Pengeluaran ---")
for x,y in laporan.items():
    print(f"{x.capitalize():<10}: {y}")
```
