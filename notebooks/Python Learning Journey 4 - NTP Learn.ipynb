```python
import ntplib

c = ntplib.NTPClient()
response = c.request('pool.ntp.org')
print(response.tx_time)
```

```python
from datetime import datetime
print(datetime.fromtimestamp(response.tx_time))
```

```python
from datetime import datetime, timezone, timedelta

wib = timezone(timedelta(hours=7))
waktu = datetime.fromtimestamp(response.tx_time, tz=wib)
print(waktu)
```

```python
def get_time():
    try:
        c = ntplib.NTPClient()
        response = c.request('pool.ntp.org')
        wib = timezone(timedelta(hours=7))
        waktu = datetime.fromtimestamp(response.tx_time, tz=wib)
        sumber = "NTP"
    except:
        waktu = datetime.now()
        sumber = "LOKAL"
        print("⚠️ Koneksi NTP gagal, menggunakan waktu lokal")
    return waktu, sumber

get_time()
```

```python
waktu, sumber = get_time()
print(waktu)
print(sumber)
```
