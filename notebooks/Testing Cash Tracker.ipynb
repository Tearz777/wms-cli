```python
transaksi = [
 {"jenis":"masuk","jumlah":500000,"kategori":"penjualan"},
 {"jenis":"keluar","jumlah":120000,"kategori":"belanja"},
 {"jenis":"keluar","jumlah":50000,"kategori":"transport"},
 {"jenis":"masuk","jumlah":200000,"kategori":"penjualan"},
 {"jenis":"keluar","jumlah":70000,"kategori":"listrik"}
]

pemasukan =0
pengeluaran=0

for i in transaksi :
    if i["jenis"] =="masuk" :
        pemasukan+= i["jumlah"]
    else :
        pengeluaran+= i["jumlah"]
    
while True :
    print("\n===Cash Tracker===")
    print("1. Lihat Saldo")
    print("2. Laporan Pemasukan")
    print("3. Laporan Pengeluaran")
    print("4. Keluar")
    try :
     com = int(input("Pilih Menu :"))
     if com==1:
         print(f"\n Saldo Terakhir : {pemasukan-pengeluaran:,}")
     elif com ==2 :
         lm = {}
         for i in transaksi:
            if i["jenis"]=="masuk" :
                x = i["kategori"]
                y = i["jumlah"]
                lm[x] = lm.get(x,0) +y
         print("\n === Laporan Pemasukan ===")
         for x,y in lm.items():
                print(f"{x.capitalize():<10}:{y: ,}")
     elif com ==3 :
        lp = {}
        for i in transaksi:
            if i["jenis"] == "keluar" :
                a = i["kategori"]
                b = i["jumlah"]
                lp[a] = lp.get(a,0) + b
        print("\n === Laporan Pengeluaran ===")
        for a,b in lp.items():
            print(f"{a.capitalize():<10}:{b:,}")
     elif com ==4 :
       print("Program Selesai")
       print("Terima Kasih Sudah Berkunjung")
       break
     else:
        print("Salah Input, Masukkan Angka Sesuai Menu")
    except ValueError:
        print("Masukkan Angka Sesuai Menu")
```
