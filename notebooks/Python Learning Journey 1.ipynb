```python

print("Mini Challenge Looping")
x=0
y=0
while True:
    angka = int(input("Masukkan Angka : "))
    if angka ==0:
        break
    elif angka >0:
      x+= angka
    else :
      y+=1
print("Total Positive: ",x)
print("Total Negative: ",y)
```

```python
print("Mini Task Max-Min")

x=None
y=None

while True:
    angka=float(input("Masukkan Angka (MASUKKAN 0 UNTUK END) :"))
    if angka ==0:
        print("Permainan Selesai")
        break
    if x is None and y is None:
        if x is None or angka >x:
         x=angka
         y=angka
         print (f"Angka Terbesar dan Terkecil adalah {x}dan {y}")
        elif angka <x:
         y=angka
         print(f"Angka Terbesar adalah :{x} dan Angka Terkecil adalah : {y}")

print("Max Value:", x)
print("Min Value:", y)
```

```python
print("=== Buku Kas Sederhana ===")

saldo = 0
infoa =[]
infob =[]
masuk=0
keluar=0
while True:
    print("\n1. Tambah Pemasukan")
    print("2. Tambah Pengeluaran")
    print("3. Lihat Saldo")
    print("4. Keluar")

    pilih = input("Pilih menu: ")

    if pilih == "1":
        info =input("Masukkan Informasi Pemasukan : ")
        masuk1=float(input("Masukkan Jumlah Pendapatan :"))
        infoa.append(f"{info}:{masuk1}")
        saldo +=masuk1
        masuk +=masuk1
        print("Pemasukan Dicatat")
        
    elif pilih == "2":
        info2 =input("Masukkan Informasi Pengeluaran :")
        keluar1=float(input("Masukkan Jumlah Pengeluaran :"))
        if saldo <keluar1:
            print("Saldo Kurang Dari Pengeluaran, cek kembali pengeluaran!")
        else :
         infob.append(f"{info2}:{keluar1}")
         saldo -=keluar1
         keluar +=keluar1
         print("Pengeluaran Berhasil Dicatat")

    elif pilih == "3":
        print("==Mini Cash Tracker==")
        print(f"Total Saldo : {saldo}")
        print(f"Pemasukan : {infoa}, {masuk}")
        print(f"Pengeluaran: {infob}, {keluar}")
        print(f"Total Pemasukan  : {masuk}")
        print(f"Total Pengeluaran: {keluar}")

    elif pilih == "4":
        print("Program selesai")
        break

    else:
        print("Pilihan tidak valid")
```

```python
print("===Mini Task - Mini Cash Tracker v2===")

transaksi = []

saldo = 0
masuk = 0
keluar = 0
while True:
    try :
        print("\n1. Pemasukan")
        print("2. Pengeluaran")
        print("3. Lihat Saldo")
        print("4. Keluar")
        pilih = int(input("Masukkan Pilihan Menu :"))
    except ValueError :
        print("Masukkan Angka Woy, Bukan Harapan💀")
        continue
    if pilih == 1:
        info = input("Masukkan Informasi Transaksi :")
        if any(char.isdigit() for char in info)or info.strip()== "":
            print("Nama Transaksinya woy!!")
        else :
           try :
               angka= int(input("Masukkan Nominal Transaksi :"))
               if angka <=0:
                   print("Yang bener woy, masa pemasukan minus atau 0💀💀")
               else :
                  t1 = {
                "Jenis" : "Pemasukan",
                "Keterangan" : info,
                "Jumlah" : angka
                     }
                  transaksi.append(t1)
                  saldo += angka
                  masuk += angka
                  print("Pemasukan Berhasil Dicatat")
           except ValueError:
               print("Nominal woy, buka doa💀💀")
  
    elif pilih ==2:
        info2 = input("Masukkan Informasi Transaksi:")
        if any(char.isdigit() for char in info2)or info2.strip()=="":
            print("Nama Transaksinya woy!!")
        else :
            try :
                angka2 = int(input("Masukkan Nominal Pengeluaran:"))
                if saldo < angka2 :
                             print("Saldo Kurang, Cek Lagi Pengeluaran!")
                else :
                   t2 = {
                    "Jenis" : "Pengeluaran",
                    "Keterangan" : info2,
                    "Jumlah" : angka2
                }
                   transaksi.append(t2)
                   saldo -= angka2
                   keluar +=angka2
                   print("Pengeluaran Berhasil Dicatat")
            except ValueError :
                print("NOMINAL ITU ANGKA!!")
   
    elif pilih ==3:
        print("==Mini Cash Tracker v2 By Thery==")
        print(f"Total Saldo : {saldo}")
        print(f"Total Pemasukan: {masuk}")
        print(f"Total Pengeluaran: {keluar}")
        print("-"*30)
        for item in transaksi:
            print(f"[{item["Jenis"]}],{item["Keterangan"]}:{item["Jumlah"]}")
 
    elif pilih ==4 :
        print("Program Selesai, Terima kasih telah Menggunakan Cash Tracker")
        break
    else :
      print("input Yang Bener Woy💀💀")
    
```

```python
transaksi = []

def hitung(transaksi):
    saldo = 0
    while True:
        info = input("Masukkan Informasi Transaksi (ketik 'exit' untuk berhenti): ")
        
        if info.lower() == 'exit':
            break
            
        try:
            angka = int(input("Masukkan Nominal : "))
            
            if angka <= 0:
                print("Pemasukan tidak boleh kurang dari atau sama dengan 0! 💀")
            else :
                t1 = {
                    "Jenis": "Masuk",
                    "Keterangan": info,
                    "Nominal": angka
                }
                transaksi.append(t1)
                saldo += angka
                print(f"OK! Berhasil dicatat. Saldo sekarang: {saldo}")
                
        except ValueError :
            print("Nominal Harus Angka Woy! Jangan masukin huruf! 💀")

hitung(transaksi)
print("\nRiwayat Transaksi Akhir:", transaksi)
```
